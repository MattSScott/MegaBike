```python
import matplotlib.pyplot as plt
import networkx as nx
import numpy as np
import os
import pandas as pd
import pickle
import json
import dit
import subprocess
from dit import Distribution
from dit.shannon import mutual_information as mi
from emergence.calc.dit import DitDistCalc
from emergence.micalc import MutualInfo
from emergence.utils.jvm import JVM
import ptitprince as pt
```

```python
# Create a dummy joint distribution representing perfect team formation
num_agents = 24  # 3 teams of 8
joint_counts = {}

# Create outcomes for each pair
for i in range(num_agents):
    for j in range(i+1, num_agents):
        # For each pair, need both possible outcomes (same bike=0 or 1)
        same_team = (i//8 == j//8)  # True if agents i,j are in same team of 8
        
        # Create both outcomes for this pair
        key_same = (i, j, 1)
        key_diff = (i, j, 0)
        
        # Add to outcomes with appropriate counts
        joint_counts[key_same] = 100 if same_team else 0
        joint_counts[key_diff] = 0 if same_team else 100

# Convert to distribution format
outcomes = list(joint_counts.keys())
total = sum(joint_counts.values())
probs = [joint_counts[o] / total for o in outcomes]

# Create distribution and calculate psi
dist = Distribution(outcomes, probs)
calc = DitDistCalc(dist, mi)
psi = calc.psi(q=2)
print(f"Psi for perfectly stable teams: {psi}")
```

```python
# Create a dummy joint distribution representing random assignments
num_agents = 24  # 3 teams of 8
joint_counts = {}

# Create outcomes for each pair
for i in range(num_agents):
    for j in range(i+1, num_agents):
        # For completely random assignment, probability is equal for same/different bike
        # Each bike has 8/24 = 1/3 chance, so P(same bike) = 1/3
        key_same = (i, j, 1)
        key_diff = (i, j, 0)
        
        # Set equal counts - using 100 as base count
        joint_counts[key_same] = 33  # approximating 1/3 * 100
        joint_counts[key_diff] = 67  # approximating 2/3 * 100

# Convert to distribution format
outcomes = list(joint_counts.keys())
total = sum(joint_counts.values())
probs = [joint_counts[o] / total for o in outcomes]

# Create distribution and calculate psi
dist = Distribution(outcomes, probs)
calc = DitDistCalc(dist, mi)
psi = calc.psi(q=2)
print(f"Psi for random assignments: {psi}")
```
